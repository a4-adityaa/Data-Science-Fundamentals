# **Matplotlib**
Matplotlib is a comprehensive, open-source Python library for creating static, animated, and interactive visualizations.

we will use `matplotlib.pyplot` a user-friendly interface for plotting

```python
# importing matplotlib
import matplotlib.pyplot as plt
```

Let's setup some data

```python
# we can also do with numpy array
x= [2023, 2024, 2025, 2026]
y= [15, 25, 30, 20]

plt.plot(x,y)
```

# Customization


```python
import numpy as np
x= np.array([2023, 2024, 2025, 2026])
y1= np.array([15, 25, 30, 20])
y2= np.array([17, 23, 38, 5])

plt.plot(x,y, marker='.',
         markersize=20,
         markerfacecolor='#1cd3fc',
         linestyle= "solid",
         linewidth= 3,
         color="#1c5bfc")
```

# Creating a line style

```python
line_style = dict(marker='.',
                  markersize=20,
                  markerfacecolor='#1cd3fc',
                  linestyle= "solid",
                  linewidth= 3)

plt.plot(x,y1, **line_style) # Passing a dictionary of style
plt.plot(x,y2, **line_style)
```

```python
y3= [13, 15, 20, 30]
plt.plot(x,y1, color= "#f86399", **line_style)
plt.plot(x,y2, color="#fda106" ,**line_style)
plt.plot(x,y3, **line_style)
```

# **Label**

```python
x= np.array([2023, 2024, 2025, 2026])
y1= np.array([15, 25, 30, 20])
y2= np.array([17, 23, 38, 5])

plt.title("Class size", fontsize= 20,
          # family="Arial",
          fontweight="bold",
          color="#2d4cfc")
plt.xlabel("Year", fontsize= 15,
                    fontstyle="italic",
                    color="#fc0000")
plt.ylabel("Students", fontsize= 15,
                    fontstyle="italic",
                    color="#00fc1c")

plt.tick_params(axis="both",
                colors="#0005fc")
plt.plot(x,y1)
plt.plot(x,y2)
plt.plot(x,y3)

plt.xticks(x)
```
