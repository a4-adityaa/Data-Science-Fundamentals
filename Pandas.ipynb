# **Pandas**
Pandas is an open-source Python library used for data manipulation and analysis, specifically designed to handle structured, tabular data efficiently.



**Importing Pandas**


We can import pandas by using keyword `import pandas as pd`

```python
import pandas as pd
```

# Series
A pandas 1-Dimensional labled array that can hold any data type...

Think of it like a single column in a spredsheet (1-Dimensional)

```python
data = [100,50,75,63,88]
series = pd.Series(data, index=["a", "b","c","d","e"])
series

```

We can use `loc[]` for accesing value using lable and `iloc[]` for accesing value using index as other programming language...

```python
# we can access the single value of a list using .loc[]
print(series.loc["a"])
```

```python
#Series is also mutable
series.loc["c"] = 200
series
```

```python
# Access using Index
print(series.iloc[2])
```

# Filtering
We can also filter data using operators

```python
# greater than 100
print(series[series >= 100])
```

Let's try with some dictionay....

```python
calories ={"Day 1": 1750, "Day 2": 2100, "Day 3": 1700}
# The key value will be auto recognized as label and we don't explicitly need to assign label
series= pd.Series(calories)
series
```

```python
# we can aslo access the values with the help of key as Python dictionary...
series.loc["Day 3"] += 500
series
```

```python
# filtering
print(series[series > 2000])
```

# **DataFrame**
A tabular data structure with rows and columns. (2-Dimensional).... similar to an excel spreadsheet.

```python
data = {"Name": ["Aditya","Ram","Ankit"],
        "Age":[30,35,50]}
df = pd.DataFrame(data)
df
```

```python
df = pd.DataFrame(data, index=["Employee 1","Employee 2","Employee 3"])
# we can aslo access data using `loc[]` keyword

print(df.loc["Employee 2"])
print(df.loc[["Employee 1"]])
```

# Adding a new Column
For adding a new column the number of rows sholud be equal



```python
df["Job"] = ["cook","N/A","Cashier"]
df
```

# Adding a new row
For adding a new row we pass a list of dictionary into DataFrame

```python
new_row= pd.DataFrame([{"Name": "Prince", "Age": 23, "Job":"Engineer"}], index=["Employee 4"])
df = pd.concat([df, new_row])
df
```
