# **Seaborn**
Seaborn is a library for making statistical graphics in `Python`. It builds on top of `matplotlib` and integrates closely with pandas data structures.

First we will install seaborn and after that import it...

Along with we will install `numpy`, `pandas` and `matplotlib`

```python
import seaborn as sns
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
```

# **Loading some data**
Here we are going to use Penguin datasets

```python
sns.get_dataset_names()
```

```python
penguins = sns.load_dataset("penguins")
penguins.head()
```

```python
# let's specify some columns
penguins["species"].value_counts()
```

Alright we got our data... now let's perform some operations to visulize it.

# Parameters and features
* **Set Style**

Set the background and appreance of graph
   
    There are five types of styles
    * white
    * Drak
    * whiteGrid
    * darkgrid
    * ticks
* **bold text**

Control overall scale of graph such as text, label etc....
    
    Generally it is of four types
    * paper
    * notebook
    * talk
    * poster

* Hue : Shows on what basis the cateogories should be diffrernt
* palette: Shows which color scheme should be used to categories

```python
sns.set_style("white")
sns.set_context("notebook") # the four types of context are (notebook, talk, poster and paper) all are used for viewing or printing the best format of graph
sns.scatterplot(data= penguins, x="flipper_length_mm",y="body_mass_g", hue= "sex", palette="Set1") # hue is extra variable to compare
# pallete defines the color
sns.despine()
# used to remove spines from graph
```

```python
sns.scatterplot(data= penguins, x="species", y= "body_mass_g", hue= "island", style= "sex", alpha= 0.5)
```

```python
sns.stripplot(data= penguins, x="species", y= "body_mass_g", hue= "island", dodge=True, jitter=True)
```

```python
sns.set_context("talk")
sns.swarmplot(data= penguins, x="species", y= "body_mass_g", hue= "island")
sns.despine()
```

```python
sns.set_context("notebook")
sns.histplot(data= penguins, x= "body_mass_g")
```
