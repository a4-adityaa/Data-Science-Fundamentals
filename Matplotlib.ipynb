# **Matplotlib**
Matplotlib is a comprehensive, open-source Python library for creating static, animated, and interactive visualizations.

we will use `matplotlib.pyplot` a user-friendly interface for plotting

```python
# importing matplotlib
import matplotlib.pyplot as plt
```

Let's setup some data

```python
# we can also do with numpy array
x= [2023, 2024, 2025, 2026]
y= [15, 25, 30, 20]

plt.plot(x,y)
```

# Customization


```python
import numpy as np
x= np.array([2023, 2024, 2025, 2026])
y1= np.array([15, 25, 30, 20])
y2= np.array([17, 23, 38, 5])

plt.plot(x,y, marker='.',
         markersize=20,
         markerfacecolor='#1cd3fc',
         linestyle= "solid",
         linewidth= 3,
         color="#1c5bfc")
```

# Creating a line style

```python
line_style = dict(marker='.',
                  markersize=20,
                  markerfacecolor='#1cd3fc',
                  linestyle= "solid",
                  linewidth= 3,
                  color="#1c5bfc")

plt.plot(x,y1, **line_style) # Passing a dictionary of style
plt.plot(x,y2, **line_style)
```
